```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.load("../terrain/small/small.npy")
```

```python
data.shape
```

```python
def set_figure_pixel_size (width, height, ppi):
    plt.figure(figsize=(width/float(ppi), height/float(ppi)), dpi=ppi)

set_figure_pixel_size(data.shape[0], data.shape[1], 166)
```

```python
# plt.figure(figsize=(data.shape[0] * 30, data.shape[1] * 30))
my_dpi = 100
fig, ax = plt.subplots()
ax.set_aspect("equal")
plt.gca().invert_yaxis()
CS = ax.contour(data, levels=range(int(np.min(data)), int(np.max(data)), 10))
ax.clabel(CS, inline=True, fontsize=8)
plt.show()
```
